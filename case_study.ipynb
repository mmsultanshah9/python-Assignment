```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

data = {
    'Country': ['USA', 'China', 'India', 'Germany', 'Brazil', 'Japan', 'Russia', 'Canada', 'Australia', 'Mexico'],
    'GDP_per_Capita': [59850, 10660, 1800, 46100, 10600, 38900, 10750, 41300, 51000, 9500],
    'Life_Expectancy': [78.54, 76.91, 69.42, 81.13, 75.88, 84.6, 73.34, 82.2, 82.5, 77.03],
    'Population(millions)': [331, 1393, 1380, 83, 213, 126, 145, 38, 26, 128],
    'Color_Grading': [59850, 10660, 1800, 46100, 10600, 38900, 10750, 41300, 51000, 9500]  # Using GDP for color
}
```

```python
df = pd.DataFrame(data)
df
```

```python
GDP_per_Capita=df.GDP_per_Capita
life = df.Life_Expectancy

# Basic scatter plot, log scale
plt.scatter(GDP_per_Capita, life)
plt.xscale('log')

plt.xlabel("GDP")
plt.ylabel("life Expectancy")
plt.title("world development")

plt.show()
```

```python
plt.scatter(GDP_per_Capita, life)

plt.xscale('log')
plt.xlabel("GDP")
plt.ylabel("life Expectancy")
plt.title("world development")

tick_val = [1000, 10000, 100000]
tick_lab = ['1k', '10k', '100k']

plt.xticks(tick_val, tick_lab)

plt.show()
```

```python
pop = [10, 20, 30, 40, 50, 90, 100, 110, 120, 130]

np_pop = np.array(pop)
plt.scatter(GDP_per_Capita, life, s=np_pop)

plt.xscale('log')
plt.xlabel("GDP")
plt.ylabel("life Expectancy")
plt.title("world development")
plt.xticks([1000, 10000, 100000], ['1k', '10k', '100k'])

plt.show()
```

```python
col = ['red','green','blue','black','yellow','red','green','blue','black','yellow']
np_pop = np.array(pop)

plt.scatter(GDP_per_Capita, life, s=np_pop, c=col)

plt.xscale('log')
plt.xlabel("GDP")
plt.ylabel("life Expectancy")
plt.title("world development")
plt.xticks([1000, 10000, 100000], ['1k', '10k', '100k'])

plt.show()
```

```python
plt.scatter(x=GDP_per_Capita, y=life, s=np_pop, c=col)

plt.xscale('log')
plt.xlabel("GDP")
plt.ylabel("life Expectancy")
plt.title("world development")
plt.xticks([1000, 10000, 100000], ['1k', '10k', '100k'])

plt.text(10000, 74, 'India')
plt.text(9000, 78, 'China')

plt.grid(True)

plt.show()
```
